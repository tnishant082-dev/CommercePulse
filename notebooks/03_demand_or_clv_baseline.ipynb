# 03 — Churn & demand baselines

Still learning — these are honest baselines, not production models.

**Questions**
- If we ignore Recency (leakage risk — label is built from it), can Frequency / Monetary / Tenure still separate Churned90?
- How does a plain LogisticRegression compare to a quick GBM experiment?
- Is seasonal-naive (lag-12) good enough as a demand stub on ~25 months?


```python
from pathlib import Path
import json

ROOT = Path("..")
m = json.loads((ROOT / "reports" / "model_metrics.json").read_text())

ch = m["churn"]
print("features used:", ch["features"])
print("base rate:", ch["base_rate"])
print()
print("LogisticRegression  AUC", ch["logistic_regression"]["auc"],
      " Acc", ch["logistic_regression"]["accuracy"])
print("GradientBoosting    AUC", ch["gradient_boosting"]["auc"],
      " Acc", ch["gradient_boosting"]["accuracy"], "  (experiment)")
print()
print("AOV MAE:", m["aov_baseline"]["mae"], "  naive:", m["aov_baseline"]["naive_mean_mae"])
print("Demand MAPE %:", m["demand_baseline"]["mape_pct"])
```

### Notes to self
- Recency was dropped on purpose — including it makes the model look smart for the wrong reason.
- TenureDays dominates GBM importance here; worth double-checking before trusting the ranking.
- Demand MAPE ~17% with lag-12 is "ok for a stub", not a forecast I'd ship.
